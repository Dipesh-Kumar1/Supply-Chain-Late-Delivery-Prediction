```python
import pandas as pd

df = pd.read_csv(
    "/content/drive/MyDrive/AI-Powered-Supply-Chain-Analytics/data/processed/DataCoSupplyChainDataset_Clean.csv"
)
```

```python
df.head()
```

```python
df["order date (DateOrders)"] = pd.to_datetime(df["order date (DateOrders)"])
df["shipping date (DateOrders)"] = pd.to_datetime(df["shipping date (DateOrders)"])
```

```python
df.head()
```

### Feature1:Shipping Delay

```python
# Create Shipping Delay feature
df["Shipping Delay"] = (
    df["Days for shipping (real)"] -
    df["Days for shipment (scheduled)"]
)
```

```python
df[
    [
        "Days for shipping (real)",
        "Days for shipment (scheduled)",
        "Shipping Delay"
    ]
].head(10)
```

```python
# Create Order Month feature
df["Order Month"] = df["order date (DateOrders)"].dt.month_name()
```

###Feature2:Order Month

```python
df[
    [
        "order date (DateOrders)",
        "Order Month"
    ]
].head(10)
```

###Feature3:Order Month Number

```python
# Create Order Month Number
df["Order Month Number"] = df["order date (DateOrders)"].dt.month
```

```python
df[
    [
        "order date (DateOrders)",
        "Order Month",
        "Order Month Number"
    ]
].head(10)
```

###Feature 4: Order Quarter

```python
# Create Order Quarter
df["Order Quarter"] = df["order date (DateOrders)"].dt.quarter
```

```python
df[
    [
        "order date (DateOrders)",
        "Order Quarter"
    ]
].head(10)
```

###Feature 5: Order Year

```python
# Create Order Year
df["Order Year"] = df["order date (DateOrders)"].dt.year
```

```python
df[
    [
        "order date (DateOrders)",
        "Order Year"
    ]
].head()
```

###Feature 6: Order Day Name

```python
# Create Order Day
df["Order Day"] = df["order date (DateOrders)"].dt.day_name()
```

```python
df[
    [
        "order date (DateOrders)",
        "Order Day"
    ]
].head()
```

###Feature 7: Day Type

```python
# Create Day Type feature
df["Day Type"] = df["Order Day"].apply(
    lambda x: "Weekend" if x in ["Saturday", "Sunday"] else "Weekday"
)
```

```python
df[
    [
        "Order Day",
        "Day Type"
    ]
].head(10)
```

###feature8:Profit Margin

```python
# Create Profit Margin feature
df["Profit Margin"] = (
    df["Order Profit Per Order"] / df["Sales"]
) * 100
```

```python
df[
    [
        "Sales",
        "Order Profit Per Order",
        "Profit Margin"
    ]
].head(10)
```

```python
df.to_csv(
    "/content/drive/MyDrive/AI-Powered-Supply-Chain-Analytics/data/processed/DataCoSupplyChainDataset_FeatureEngineered.csv",
    index=False
)
```
